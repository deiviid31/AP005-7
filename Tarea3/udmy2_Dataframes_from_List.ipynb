Las siguientes líneas son para **indicar mediante un comentario el tema del notebook (crear un DataFrame a partir de una lista) e importar la librería pandas**.

```python
# Create dataframe from a list

import pandas as pd
```

La siguiente línea es para **crear una lista simple de números**.

```python
numbers = [1, 2, 3, 4, 5, 6]
```

La siguiente línea es para **crear un DataFrame de pandas a partir de la lista de números**.

```python
pd.DataFrame(numbers)
```

La siguiente línea es para **crear una lista de listas, donde cada sublista representa una fila con nombre y edad**.

```python
data = [["Jhon",  23], ["Mary", 18], ["Jhon", 31]]
```

La siguiente línea es para **crear un DataFrame de pandas a partir de la lista de listas**.

```python
pd.DataFrame(data)
```

La siguiente línea es para **crear el DataFrame indicando los nombres de las columnas ("name" y "age")**.

```python
pd.DataFrame(data, columns = ["name", "age"])
```

La siguiente línea es para **crear el DataFrame indicando tanto los nombres de las columnas como un índice personalizado (id1, id2, id3)**.

```python
pd.DataFrame(data, columns = ["name", "age"], index = ["id1", "id2", "id3"])
```
