Las siguientes líneas son para **importar pandas y leer el archivo Excel "Nba Teams.xlsx" para crear el DataFrame, mostrando las primeras 5 filas**.

```python
import pandas as pd
nba = pd.read_excel("Nba Teams.xlsx")
nba.head()
```

Las siguientes líneas son para **crear dos nuevas columnas vacías ("Teams" y "nba Teams") en el DataFrame**.

```python
nba["Teams"] = " "
nba["nba Teams"] = " "
nba.head()
```

Las siguientes líneas son para **acceder a la columna "Teams" usando notación de atributo (punto); las líneas comentadas muestran que esto no funciona con columnas cuyo nombre tiene espacios o que no existen**.

```python
#nba.Name
#nba.nba Teams
nba.Teams
#nba.Sports
```

Las siguientes líneas son para **crear una nueva columna "Sports" y asignarle el valor "basket" a todas las filas**.

```python
nba["Sports"] = "basket"
nba.head()
```

Las siguientes líneas son para **eliminar la columna "Teams" de forma permanente con inplace=True (las líneas comentadas muestran alternativas con insert() que no se ejecutan)**.

```python
nba.drop("Teams", axis = 1, inplace = True)
#nba.insert(2, column = "Teams", value = "Colombians")
#nba.insert(2, column = "Sportss", value = "Turmequé")
nba
```
